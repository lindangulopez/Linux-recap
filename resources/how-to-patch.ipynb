This notebook demonstrates what **Git patches** are, how to create them, and how they can be preserved and referenced in **Software Heritage**.

## What is a Git Patch?

- A **patch** is a file that describes changes made to a codebase.
- Shows **removed lines** with `-` and **added lines** with `+`.
- Can be shared and applied without needing direct access to a repository.
- Useful for code review, email-based collaboration, and **archiving development history**.

## Step 3 — Generate a Git Patch

We can export our last commit as a **patch file** that captures all changes and metadata.

````python
# Illustrative command
# !git format-patch -1 HEAD

from IPython.display import display, HTML

# Example patch filename (simulated)
patch_file = '0001-Fix-typo-in-codemeta.json.patch'

display(HTML(f'<b>Generated patch file:</b> {patch_file}'))
````

### What the Patch Contains

- Commit metadata: author, date, commit message
- Diff of changes: lines removed (-) and lines added (+)
- Can be applied to a repository using `git am`

## Step 4 — Share and Apply a Patch

The patch can be emailed, uploaded, or shared. Another developer can apply it with:

```bash
git am 0001-Fix-typo-in-codemeta.json.patch
```

## Step 5 — Linking to Software Heritage

Once archived, the patch and original files can be referenced permanently. Here are example clickable links for the **codemeta/codemeta** repository and files:

````python
GITHUB_REPO = "https://github.com/codemeta/codemeta"

swhid_objects = {
    'Root directory': 'swh:1:dir:a0a98035d99358c55b96731583a892fe06c60528',
    'codemeta.json': 'swh:1:cnt:9b2c3b2fa1a2c87b5a1f7d1e5a7f7fbb8d1a2345',
    'Patch example': 'swh:1:cnt:PATCH_PLACEHOLDER'
}

for name, swhid in swhid_objects.items():
    url = f'https://archive.softwareheritage.org/browse/origin/{GITHUB_REPO}/?swhid={swhid}'
    display(HTML(f"<a href='{url}' target='_blank'>{name} (View in Software Heritage)</a><br>"))
````

### Notes:
- Replace `PATCH_PLACEHOLDER` with the actual SWHID once the patch is archived.
- This allows **students and researchers** to explore the patch content interactively.

## ✅ Summary

- Git patches are **portable, reviewable artifacts**.
- They allow collaboration without direct repository access.
- Once archived, patches and files can be **referenced permanently using SWHIDs**.
- This approach is perfect for **teaching, reproducible research, and long-term software preservation**.
